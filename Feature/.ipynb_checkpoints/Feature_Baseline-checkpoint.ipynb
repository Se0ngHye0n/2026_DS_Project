# Baseline Feature Engineering
## 기본 feature engineering

```python
from pathlib import Path

import pandas as pd
import numpy as np
```

```python
# ================
# Raw 데이터 로드
# ================
# 데이터 로드
DATA_DIR = Path('../Data')
df_raw = pd.read_csv(DATA_DIR / 'train_2025.csv')
```

```python
# ====================
# 1️. 기본 Numeric Raw
# ====================
raw_numeric = [
    'age_of_driver',
    'safty_rating',
    'annual_income',
    'past_num_of_claims',
    'liab_prct',
    'claim_est_payout',
    'age_of_vehicle',
    'vehicle_price',
    'vehicle_weight'
]
```

```python
# =============
# 2️. 단순 파생
# =============
df_raw['log_claim_est_payout'] = np.log1p(df_raw['claim_est_payout'])
df_raw['log_annual_income'] = np.log1p(df_raw['annual_income'])

df_raw['payout_income_ratio'] = (
    df_raw['claim_est_payout'] /
    (df_raw['annual_income'] + 1)
)

simple_derived = [
    'log_claim_est_payout',
    'log_annual_income',
    'payout_income_ratio'
]
```

```python
# =========================
# 3️. Raw Binary Behavioral
# =========================
raw_binary = [
    'witness_present_ind',
    'policy_report_filed_ind',
    'address_change_ind',
    'high_education_ind'
]
```

```python
# ===================
# 4️. Raw Categorical
# ===================
raw_categorical = [
    'gender',
    'marital_status',
    'living_status',
    'accident_site',
    'channel',
    'vehicle_category',
    'vehicle_color',
    'claim_day_of_week'
]
```

```python
# ====================================
# 5️. 최종 Raw Baseline Feature 리스트
# ====================================
baseline_features = (
    raw_numeric +
    simple_derived +
    raw_binary +
    raw_categorical
)
```

```python
# =================================
# 6️. Baseline Feature Dataset 생성
# =================================
X_raw_baseline = df_raw[baseline_features].copy()
y_raw_baseline = df_raw['fraud'].copy()

print("Raw Baseline Feature Shape:", X_raw_baseline.shape)
print("Target Shape:", y_raw_baseline.shape)

display(X_raw_baseline.head())
```

```python
# ===========================
# 7. Baseline Feature Set 저장
# ===========================
raw_baseline_df = X_raw_baseline.copy()
raw_baseline_df['fraud'] = y_raw_baseline.values

DATA_DIR = Path('../Data')
raw_baseline_df.to_csv(DATA_DIR / 'train_baseline.csv', index=False)
```
