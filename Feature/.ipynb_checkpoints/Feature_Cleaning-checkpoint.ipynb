```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from sklearn.model_selection import StratifiedKFold
```

```python
# ===============
# 1. 데이터 로드
# ===============
DATA_DIR = Path('../Data')

train = pd.read_csv(DATA_DIR / 'train_V4.csv')
test  = pd.read_csv(DATA_DIR / 'test_V4.csv')

# 원본 데이터에서 누락된 카테고리 피처 가져오기
train_raw = pd.read_csv(DATA_DIR / 'train_2025.csv')
test_raw  = pd.read_csv(DATA_DIR / 'test_2025.csv')

print(f"Train V4: {train.shape}  |  Test V4: {test.shape}")
print(f"Fraud rate: {train['fraud'].mean():.2%}")
```

```python
# ============================================
# 2. 순서 보존 확인 후 원본 카테고리 피처 추가
# ============================================
assert (train_raw['age_of_driver'].values == train['age_of_driver'].values).all(), \
    "Train 행 순서 불일치!"
assert (test_raw['age_of_driver'].values == test['age_of_driver'].values).all(), \
    "Test 행 순서 불일치!"

# gender 추가 (V4에서 누락)
train['gender'] = train_raw['gender'].values
test['gender']  = test_raw['gender'].values

print("Gender 분포 (train):")
print(train['gender'].value_counts())
print("\nGender별 사기율:")
print(train.groupby('gender')['fraud'].mean())
```

```python
# ====================================
# 10. 저중요도 피처 제거 (노이즈 감소)
# ====================================
# V4 모델에서 importance = 0 이거나 매우 낮은 피처들
low_importance_cols = [
    'claim_bin',
    'veh_size_nan',
    'payout_to_income_ratio_clipped_top1_flag',
    'high_payout_flag',
    'old_vehicle_flag',
    'payout_to_residual_ratio_clipped_top5_flag',
    'payout_to_residual_ratio_clipped_top1_flag',
    'over_value_claim_flag',
    'payout_to_price_ratio_clipped_top1_flag',
    'payout_to_income_ratio_clipped_top5_flag',
    'extreme_payout_flag',
    'is_month_end_claim',
    'is_year_end_claim',
    'high_income_ratio_and_no_report',
    'high_residual_ratio_and_repeat',
    'unsafe_full_liability',
    'young_expensive_vehicle',
    'expensive_vehicle_flag',
    'overpriced_vehicle_flag',
    'zero_payout_claim',
    'payout_to_price_ratio_clipped_top5_flag',
]

dropped = [c for c in low_importance_cols if c in train.columns]
train = train.drop(columns=dropped)
test  = test.drop(columns=[c for c in low_importance_cols if c in test.columns])

print(f"제거된 피처 ({len(dropped)}개): {dropped}")

# gender 문자열 컬럼 제거
train = train.drop(columns=['gender'])
test  = test.drop(columns=['gender'])

print(f"Train shape after drop: {train.shape}")
```

```python
# ======================
# 11. 최종 확인 및 저장
# ======================
X_train = train.drop(columns=['fraud']).copy()
y_train = train['fraud']
X_test  = test

print(f"X_train: {X_train.shape}")
print(f"y_train: {y_train.shape}  (fraud={y_train.mean():.2%})")
print(f"X_test:  {X_test.shape}")

# 신규 추가 피처 목록
v4_cols = set(pd.read_csv(DATA_DIR / 'train_V4.csv').columns)
new_cols = [c for c in X_train.columns if c not in v4_cols]

# 저장
X_train.assign(fraud=y_train.values).to_csv(DATA_DIR / 'train_cleaning.csv', index=False)
X_test.to_csv(DATA_DIR / 'test_cleaning.csv', index=False)
```
